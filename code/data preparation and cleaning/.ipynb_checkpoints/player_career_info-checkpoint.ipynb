```python
# Loading raw player info data
import pandas as pd
from pathlib import Path
from nba_api.stats.endpoints import CommonAllPlayers

# Downloading all historical NBA players
players_info = CommonAllPlayers(
    is_only_current_season=0,
    season="2025-26",
    league_id="00"
).get_data_frames()[0]

# Saving raw data
raw_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

output_file = raw_dir / "players_info_raw.csv"
players_info.to_csv(output_file, index=False)

print("Saved raw player info data to:", output_file)
print("Original players_info shape:", players_info.shape)

players_info.head(5)
```

```python
# Cleaning player career info

players_info_clean = players_info.copy()

# Converting career years
players_info_clean["FROM_YEAR"] = pd.to_numeric(players_info_clean["FROM_YEAR"], errors="coerce")
players_info_clean["TO_YEAR"] = pd.to_numeric(players_info_clean["TO_YEAR"], errors="coerce")

# Filtering relevant players
players_info_filtered = players_info_clean[
    players_info_clean["TO_YEAR"] >= 1996
].copy()

# Selecting useful variables
player_career_info = players_info_filtered[[
    "PERSON_ID",
    "DISPLAY_FIRST_LAST",
    "FROM_YEAR",
    "TO_YEAR",
    "ROSTERSTATUS"
]].copy()

# Renaming columns
player_career_info = player_career_info.rename(columns={
    "PERSON_ID":"player_id",
    "DISPLAY_FIRST_LAST":"player_name",
    "FROM_YEAR":"career_start_year",
    "TO_YEAR":"career_end_year",
    "ROSTERSTATUS":"roster_status"
})

# Cleaning career years
player_career_info["career_start_year"] = player_career_info["career_start_year"].astype("Int64")
player_career_info["career_end_year"] = player_career_info["career_end_year"].astype("Int64") + 1
player_career_info["roster_status"] = player_career_info["roster_status"].astype("Int64")

# Correcting Mark Jones career start year
player_career_info.loc[
    player_career_info["player_id"] == 2891,
    "career_start_year"
] = 2004

# Reordering columns
player_career_info = player_career_info[[
    "player_id",
    "player_name",
    "career_start_year",
    "career_end_year",
    "roster_status"
]]

# Saving cleaned data
processed_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

output_file = processed_dir / "player_career_info_clean.csv"
player_career_info.to_csv(output_file, index=False)

print("Saved cleaned player career info data to:", output_file)
print("Original players_info rows:", players_info.shape[0])
print("Final player_career_info shape:", player_career_info.shape)

player_career_info.head()
```
