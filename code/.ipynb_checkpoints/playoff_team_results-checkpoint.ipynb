```python
############################################# create official playoff teams and seeds from playoff series history #############################################

import pandas as pd

# Load Basketball-Reference playoff series history file
file_path = r"C:\Users\ahmed\Desktop\NBA_Project\data\raw\playoff_series_history.xls"

# Basketball-Reference downloads are often HTML tables, even if the file ends with .xls
tables = pd.read_html(file_path)

playoff_series = tables[0].copy()

# Keep useful columns
playoff_series_clean = playoff_series.iloc[:, [0, 1, 2, 3, 5, 6, 8, 9]].copy()

playoff_series_clean.columns = [
    "year",
    "league",
    "series",
    "date_range",
    "winner_team",
    "winner_wins",
    "loser_team",
    "loser_wins"
]

# Keep NBA seasons from 1998 to 2026
playoff_series_clean = playoff_series_clean[
    (playoff_series_clean["league"] == "NBA") &
    (playoff_series_clean["year"] >= 1998) &
    (playoff_series_clean["year"] <= 2026)
].copy()
playoff_series_clean
```

```python
############################################# create playoff team results from playoff series data ##########################################

import pandas as pd

# Work on a copy of the playoff series data
playoff_series_work = playoff_series_clean.copy()

```

```python
############################################# extract playoff seeds from team names #############################################

# Extract playoff seed from team names
# Example: "Utah Jazz (1)" becomes seed 1
playoff_series_work["winner_seed"] = (
    playoff_series_work["winner_team"]
    .astype(str)
    .str.extract(r"\((\d+)\)")
    .astype("Int64")
)

playoff_series_work["loser_seed"] = (
    playoff_series_work["loser_team"]
    .astype(str)
    .str.extract(r"\((\d+)\)")
    .astype("Int64")
)
```

```python
############################################# remove playoff seeds from team names #############################################

# Remove playoff seed from team names
# Example: "Utah Jazz (1)" becomes "Utah Jazz"
playoff_series_work["winner_team"] = (
    playoff_series_work["winner_team"]
    .astype(str)
    .str.replace(r"\s*\(\d+\)", "", regex=True)
    .str.strip()
)

playoff_series_work["loser_team"] = (
    playoff_series_work["loser_team"]
    .astype(str)
    .str.replace(r"\s*\(\d+\)", "", regex=True)
    .str.strip()
)
```

```python
############################################# add conference information #############################################

# The conference can be extracted from the series name
# Finals rows do not contain East/West, but those teams also appear in earlier conference rounds
def get_conference_from_series(series_name):
    if "Eastern" in str(series_name):
        return "East"
    elif "Western" in str(series_name):
        return "West"
    else:
        return pd.NA

playoff_series_work["conference"] = playoff_series_work["series"].apply(get_conference_from_series)
```

```python
############################################# create conference lookup table #############################################

# Winner conference rows
winner_conference = playoff_series_work[
    [
        "year",
        "winner_team",
        "conference"
    ]
].copy()

winner_conference = winner_conference.rename(columns={
    "winner_team": "team_full_name"
})

# Loser conference rows
loser_conference = playoff_series_work[
    [
        "year",
        "loser_team",
        "conference"
    ]
].copy()

loser_conference = loser_conference.rename(columns={
    "loser_team": "team_full_name"
})

# Combine winner and loser conference rows
team_conference_lookup = pd.concat(
    [winner_conference, loser_conference],
    ignore_index=True
)

# Remove Finals rows where conference is missing
team_conference_lookup = team_conference_lookup.dropna(subset=["conference"])

# Keep one conference value per team-season
team_conference_lookup = team_conference_lookup.drop_duplicates(
    subset=["year", "team_full_name"]
)
```

```python
############################################# create playoff stage for teams that lost #############################################

def get_loser_playoff_stage(series_name):
    if series_name == "Finals":
        return "Finals Runner-up"
    elif "Conf Finals" in str(series_name):
        return "Conference Finals"
    elif "Semifinals" in str(series_name):
        return "Conference Semifinals"
    elif "First Round" in str(series_name):
        return "First Round"
    else:
        return "Unknown"

# A team receives its final playoff stage in the round where it lost
loser_results = playoff_series_work[
    [
        "year",
        "loser_team",
        "loser_seed",
        "series"
    ]
].copy()

loser_results = loser_results.rename(columns={
    "loser_team": "team_full_name",
    "loser_seed": "playoff_seed"
})

loser_results["playoff_stage"] = loser_results["series"].apply(get_loser_playoff_stage)
```

```python
############################################# create champion rows #############################################

# The Finals winner is the NBA champion
champion_results = playoff_series_work[
    playoff_series_work["series"] == "Finals"
][
    [
        "year",
        "winner_team",
        "winner_seed",
        "series"
    ]
].copy()

champion_results = champion_results.rename(columns={
    "winner_team": "team_full_name",
    "winner_seed": "playoff_seed"
})

champion_results["playoff_stage"] = "Champion"
```

```python
############################################# combine loser results and champion results #############################################

playoff_team_results = pd.concat(
    [
        loser_results[
            [
                "year",
                "team_full_name",
                "playoff_seed",
                "playoff_stage"
            ]
        ],
        champion_results[
            [
                "year",
                "team_full_name",
                "playoff_seed",
                "playoff_stage"
            ]
        ]
    ],
    ignore_index=True
)
```

```python
############################################# add conference to playoff team results #############################################

# Add conference information
playoff_team_results = playoff_team_results.merge(
    team_conference_lookup,
    on=["year", "team_full_name"],
    how="left"
)
```

```python
############################################# add season and playoff indicators #############################################

# Add season string
playoff_team_results["season"] = (
    (playoff_team_results["year"] - 1).astype(str)
    + "-"
    + playoff_team_results["year"].astype(str).str[-2:]
)

# This dataset only contains teams that actually played in the playoffs
playoff_team_results["actual_playoff_team"] = 1

# All teams in this dataset reached the playoff field
playoff_team_results["top_8_conference_team"] = 1

# Champion indicator
playoff_team_results["won_nba_title"] = (
    playoff_team_results["playoff_stage"] == "Champion"
).astype(int)

# Finals indicator
playoff_team_results["reached_finals"] = (
    playoff_team_results["playoff_stage"].isin([
        "Champion",
        "Finals Runner-up"
    ])
).astype(int)

# Conference Finals indicator
playoff_team_results["reached_conference_finals"] = (
    playoff_team_results["playoff_stage"].isin([
        "Champion",
        "Finals Runner-up",
        "Conference Finals"
    ])
).astype(int)

# Conference Semifinals indicator
playoff_team_results["reached_conference_semifinals"] = (
    playoff_team_results["playoff_stage"].isin([
        "Champion",
        "Finals Runner-up",
        "Conference Finals",
        "Conference Semifinals"
    ])
).astype(int)
```

```python
############################################# add numeric playoff finish rank #############################################

# Numeric playoff finish rank
# Lower value means better playoff result
playoff_stage_rank_map = {
    "Champion": 1,
    "Finals Runner-up": 2,
    "Conference Finals": 3,
    "Conference Semifinals": 4,
    "First Round": 5
}

playoff_team_results["playoff_finish_rank"] = (
    playoff_team_results["playoff_stage"].map(playoff_stage_rank_map)
).astype("Int64")
```

```python
############################################# reorder columns #############################################

playoff_team_results = playoff_team_results[
    [
        "year",
        "season",
        "team_full_name",
        "conference",
        "playoff_seed",
        "actual_playoff_team",
        "top_8_conference_team",
        "playoff_stage",
        "playoff_finish_rank",
        "reached_conference_semifinals",
        "reached_conference_finals",
        "reached_finals",
        "won_nba_title"
    ]
]

# Sort results
playoff_team_results = playoff_team_results.sort_values(
    [
        "year",
        "conference",
        "playoff_seed",
        "playoff_finish_rank",
        "team_full_name"
    ]
).copy()

display(playoff_team_results.head(50))
```

```python
############################################# save playoff team results #############################################

playoff_team_results.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\playoff_team_results_clean.csv",
    index=False
)
```
