```python
########################## player bio data loading ############################################

import pandas as pd
import time
from pathlib import Path
from nba_api.stats.endpoints import LeagueDashPlayerBioStats

# Function to create NBA season format
# Example: 1998 -> 1997-98
def make_season_string(end_year):
    return f"{end_year - 1}-{str(end_year)[-2:]}"

all_bio_stats = []

# Load NBA player bio data from 1997-98 to 2025-26
for end_year in range(1998, 2027):
    season = make_season_string(end_year)
    print("Loading bio stats:", season)
    
    try:
        bio = LeagueDashPlayerBioStats(
            season=season,
            season_type_all_star="Regular Season",
            per_mode_simple="Totals",
            timeout=60
        ).get_data_frames()[0]
        
        bio["Season"] = season
        bio["Year"] = end_year
        
        all_bio_stats.append(bio)
        
        time.sleep(2)
        
    except Exception as e:
        print("Could not load:", season)
        print(e)

bio_stats_all = pd.concat(all_bio_stats, ignore_index=True)

# Save raw bio data
bio_stats_all.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\raw\bio_stats_all_raw.csv",
    index=False
)
print("Raw bio data shape:", bio_stats_all.shape)
bio_stats_all.head()
```

```python
bio_stats_all.columns.tolist()
```

```python
########################## clean player bio data ############################################

# Keep useful columns from bio_stats_all
# Basic performance variables GP, PTS, REB, and AST are also kept here
# so this bio dataset can be analyzed independently if needed.
bio_info = bio_stats_all[
    [
        # Merge / identification variables
        "PLAYER_ID",
        "PLAYER_NAME",
        "TEAM_ID",
        "TEAM_ABBREVIATION",
        "Season",
        "Year",
        # Player background variables
        "AGE",
        "PLAYER_HEIGHT_INCHES",
        "PLAYER_WEIGHT",
        "COLLEGE",
        "COUNTRY",
        "DRAFT_YEAR",
        "DRAFT_ROUND",
        "DRAFT_NUMBER",

        # Basic performance variables
        "GP",
        "PTS",
        "REB",
        "AST",

        # Useful advanced performance variables
        "NET_RATING",
        "OREB_PCT",
        "DREB_PCT",
        "USG_PCT",
        "TS_PCT",
        "AST_PCT"
    ]
].copy()

# Convert numeric columns from text to numbers
bio_info["PLAYER_HEIGHT_INCHES"] = pd.to_numeric(
    bio_info["PLAYER_HEIGHT_INCHES"],
    errors="coerce"
)

bio_info["PLAYER_WEIGHT"] = pd.to_numeric(
    bio_info["PLAYER_WEIGHT"],
    errors="coerce"
)

bio_info["AGE"] = pd.to_numeric(
    bio_info["AGE"],
    errors="coerce"
)

# Convert height and weight to easier units
bio_info["height_cm"] = (bio_info["PLAYER_HEIGHT_INCHES"] * 2.54).round(1)
bio_info["weight_kg"] = (bio_info["PLAYER_WEIGHT"] * 0.453592).round(1)

# Rename columns to cleaner names
bio_info = bio_info.rename(columns={
    "PLAYER_ID": "player_id",
    "PLAYER_NAME": "player_name",
    "TEAM_ID": "team_id",
    "TEAM_ABBREVIATION": "team",
    "Season": "season",
    "Year": "year",
    "AGE": "age",
    "COLLEGE": "college",
    "COUNTRY": "country",
    "DRAFT_YEAR": "draft_year",
    "DRAFT_ROUND": "draft_round",
    "DRAFT_NUMBER": "draft_number",
    "GP": "games_played",
    "PTS": "points",
    "REB": "rebounds",
    "AST": "assists",
    "NET_RATING": "net_rating",
    "OREB_PCT": "offensive_rebound_pct",
    "DREB_PCT": "defensive_rebound_pct",
    "USG_PCT": "usage_pct",
    "TS_PCT": "true_shooting_pct",
    "AST_PCT": "assist_pct"
})

# Remove .0 from age
bio_info["age"] = bio_info["age"].astype("Int64")

# Reordering columns nicely
bio_info = bio_info[
    [
        "year",
        "season",
        "player_id",
        "player_name",
        "team_id",
        "team",
        "age",
        "height_cm",
        "weight_kg",
        "college",
        "country",
        "draft_year",
        "draft_round",
        "draft_number",
        "games_played",
        "points",
        "rebounds",
        "assists",
        "net_rating",
        "usage_pct",
        "true_shooting_pct",
        "assist_pct",
        "offensive_rebound_pct",
        "defensive_rebound_pct"
    ]
]

print("Original bio_stats_all shape:", bio_stats_all.shape)
print("Clean bio_info shape:", bio_info.shape)

bio_info.head(50)
bio_info.columns.tolist()
```

```python
############################################# correct wrong draft information in bio_info #################

# Make draft columns text/object first
bio_info["draft_year"] = bio_info["draft_year"].astype("object")
bio_info["draft_round"] = bio_info["draft_round"].astype("object")
bio_info["draft_number"] = bio_info["draft_number"].astype("object")

# Correct draft information by player_id
# Jerry Smith, player_id 202536
bio_info.loc[
    bio_info["player_id"] == 202536,
    ["draft_year", "draft_round", "draft_number"]
] = [2010, "Undrafted", "Undrafted"]

# Walker Russell, player_id 201041
bio_info.loc[
    bio_info["player_id"] == 201041,
    ["draft_year", "draft_round", "draft_number"]
] = [2006, "Undrafted", "Undrafted"]

# Mark Jones, player_id 2891
bio_info.loc[
    bio_info["player_id"] == 2891,
    ["draft_year", "draft_round", "draft_number"]
] = ["Undrafted", "Undrafted", "Undrafted"]
# Save cleaned bio data
bio_info.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\bio_info_clean.csv",
    index=False
)
```
