```python
# Loading raw team standings data
import pandas as pd
import time
from pathlib import Path
from nba_api.stats.endpoints import LeagueStandingsV3

all_team_standings = []

def make_season_string(end_year):
    return f"{end_year - 1}-{str(end_year)[-2:]}"

# Downloading all seasons
for end_year in range(1997, 2027):
    season = make_season_string(end_year)
    print("Loading team standings:", season)
    try:
        standings = LeagueStandingsV3(
            league_id="00",
            season=season,
            season_type="Regular Season",
            timeout=60
        )
        season_df = standings.get_data_frames()[0]
        season_df["Season"] = season
        season_df["Year"] = end_year
        all_team_standings.append(season_df)
        time.sleep(2)
    except Exception as e:
        print("Could not load:", season, e)

# Combining all seasons
team_standings_raw = pd.concat(all_team_standings, ignore_index=True)

# Saving raw data
raw_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

output_file = raw_dir / "team_standings_raw.csv"
team_standings_raw.to_csv(output_file, index=False)

print("Saved raw team standings data to:", output_file)
print("Raw team standings shape:", team_standings_raw.shape)

team_standings_raw.head(50)
```

```python
team_standings_raw.columns.tolist()
```

```python
# Cleaning team standings data

team_standings = team_standings_raw[[
    "Year","Season","TeamID","TeamCity","TeamName",
    "Conference","Division","WINS","LOSSES","WinPCT",
    "PointsPG","OppPointsPG","DiffPointsPG"
]].copy()

# Renaming columns
team_standings = team_standings.rename(columns={
    "Year":"year",
    "Season":"season",
    "TeamID":"team_id",
    "TeamCity":"team_city",
    "TeamName":"team_name",
    "Conference":"conference",
    "Division":"division",
    "WINS":"team_wins",
    "LOSSES":"team_losses",
    "WinPCT":"team_win_pct",
    "PointsPG":"team_points_per_game",
    "OppPointsPG":"team_opp_points_per_game",
    "DiffPointsPG":"team_point_diff_per_game"
})

# Creating full team name
team_standings["team_full_name"] = (
    team_standings["team_city"].astype(str) + " " +
    team_standings["team_name"].astype(str)
)

# Reordering columns
team_standings = team_standings[[
    "year",
    "season",
    "team_id",
    "team_full_name",
    "conference",
    "division",
    "team_wins",
    "team_losses",
    "team_win_pct",
    "team_points_per_game",
    "team_opp_points_per_game",
    "team_point_diff_per_game"
]]

# Saving cleaned data
processed_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

output_file = processed_dir / "team_standings_clean.csv"
team_standings.to_csv(output_file, index=False)

print("Saved cleaned team standings data to:", output_file)
print("Raw team standings shape:", team_standings_raw.shape)
print("Clean team standings shape:", team_standings.shape)

team_standings.head(50)
```
