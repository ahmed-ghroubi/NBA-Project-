```python
from pathlib import Path
import pandas as pd
import numpy as np
import warnings
from sklearn.model_selection import train_test_split,StratifiedKFold,cross_val_score
from sklearn.linear_model import LogisticRegression,LogisticRegressionCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import roc_auc_score,accuracy_score

warnings.filterwarnings("ignore",category=FutureWarning)

NBA_file_path=(Path.home()/"Desktop"/"NBA_Project"/"data"/"processed"/"nba_final_player_season_data.csv")
NBA_df=pd.read_csv(NBA_file_path)

target_variable="actual_playoff_team"

team_performance_vars=[
    "team_wins","team_losses","team_win_pct",
    "wins","losses","win_pct",
    "team_point_diff_per_game",
    "team_points_per_game",
    "team_opp_points_per_game"
]

on_court_vars=["plus_minus","net_rating"]

identifier_vars=[
    "team_id","year","season","team","team_full_name",
    "conference","division","player_id","player_name",
    "roster_status","career_start_year","career_end_year",
    "college","country","draft_year","draft_round","draft_number"
]

excluded_vars=[
    target_variable
]+team_performance_vars+on_court_vars+identifier_vars+[
    # Remove percentage variables when corresponding counting variables exist and variables that are combination of already present variables.
    "field_goal_pct","three_point_pct","free_throw_pct", "two_point_pct",
    "assist_pct", "offensive_rebound_pct","defensive_rebound_pct","true_shooting_pct",
    "field_goal_attempts","field_goal_made","nba_fantasy_points","Points"
]

player_features=[
    column for column in NBA_df.select_dtypes(include=np.number).columns
    if column not in excluded_vars and NBA_df[column].std()>0
]

print(f"Original rows: {len(NBA_df):,}")
print(f"Numeric candidates before deletion: {len(player_features)}")
print(player_features)
```

```python
count_vars=[
    "three_points_made","three_point_attempts",
    "free_throws_made","free_throw_attempts",
    "offensive_rebounds","defensive_rebounds","rebounds",
    "assists","turnovers","steals","blocks","blocks_against",
    "personal_fouls","personal_fouls_drawn",
    "double_doubles","triple_doubles",
    "two_points_made","two_point_attempts"
]

rate_vars=["usage_pct"]

trait_vars=["age","player_experience_years","height_cm","weight_kg","team_count"]

count_vars=[variable for variable in count_vars if variable in player_features]
rate_vars=[variable for variable in rate_vars if variable in player_features]
trait_vars=[variable for variable in trait_vars if variable in player_features]

def weighted_mean(values,weights):
    valid_mask=values.notna() & weights.notna() & (weights>0)
    return np.average(values[valid_mask],weights=weights[valid_mask]) if valid_mask.any() else np.nan

team_data=[]

for (year,team_id),group in NBA_df.groupby(["year","team_id"]):
    games=(group["team_wins"]+group["team_losses"]).max()
    team_row={
    "year": year,
    "team_id": team_id,
    "team_full_name": group["team_full_name"].iloc[0],
    target_variable: int(group[target_variable].iloc[0])
    }
    team_row.update({f"{variable}_per_game":group[variable].sum()/games for variable in count_vars})
    team_row.update({f"avg_{variable}":weighted_mean(group[variable],group["minutes"]) for variable in rate_vars+trait_vars})
    team_data.append(team_row)

team=pd.DataFrame(team_data)
final_vars=[
    column for column in team.columns 
    if column not in ["year","team_id","team_full_name",target_variable]
]
print(f"Team-seasons: {len(team)}   Predictors: {len(final_vars)}")
print(final_vars)
```

```python
################################## Logistic Regression and Regularized Extension #############################################
```

```python
# Preparing data
features=team[final_vars].replace([np.inf,-np.inf],np.nan)
target=team[target_variable].astype(int)
model_data=pd.concat([features,target],axis=1).dropna()
features,target=model_data[final_vars],model_data[target_variable]

# Splitting data into training and test sets
X_tr,X_te,y_tr,y_te=train_test_split(features,target,test_size=0.3,random_state=42,stratify=target)

# Setting up stratified cross-validation
cv_strategy=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

# Creating classification models
classification_models={
    "Logistic regression":make_pipeline(StandardScaler(),LogisticRegression(penalty=None,max_iter=5000,random_state=42)),
    "Lasso logistic (L1)":make_pipeline(StandardScaler(),LogisticRegressionCV(Cs=20,cv=cv_strategy,penalty="l1",solver="saga",scoring="roc_auc",max_iter=10000,random_state=42,n_jobs=-1)),
    "Ridge logistic (L2)":make_pipeline(StandardScaler(),LogisticRegressionCV(Cs=20,cv=cv_strategy,penalty="l2",solver="lbfgs",scoring="roc_auc",max_iter=10000,n_jobs=-1))
}

# Fitting and evaluating models
results_rows=[]
trained_models={}

for model_name,classifier in classification_models.items():
    cv_scores=cross_val_score(classifier,X_tr,y_tr,scoring="roc_auc",cv=cv_strategy,n_jobs=-1)
    classifier.fit(X_tr,y_tr)
    probabilities,predictions=classifier.predict_proba(X_te)[:,1],classifier.predict(X_te)
    results_rows.append([model_name,cv_scores.mean(),cv_scores.std(),roc_auc_score(y_te,probabilities),accuracy_score(y_te,predictions)])
    trained_models[model_name]=classifier

# Creating results table
results=pd.DataFrame(results_rows,columns=["model","CV AUC mean","CV AUC sd","test AUC","test accuracy"]).set_index("model")
display(results.round(4))
```

```python
lasso_model=trained_models["Lasso logistic (L1)"].named_steps["logisticregressioncv"]

coefficient_table=pd.DataFrame({
    "predictor":final_vars,
    "coefficient":lasso_model.coef_[0]
})

coefficient_table["absolute_coeffSSicient"]=coefficient_table["coefficient"].abs()
coefficient_table=coefficient_table.sort_values("absolute_coefficient",ascending=False)

display(coefficient_table.round(4))
```

```python
lambda_rows=[]

for model_name in ["Lasso logistic (L1)","Ridge logistic (L2)"]:
    logistic_model=trained_models[model_name].named_steps["logisticregressioncv"]
    selected_c=logistic_model.C_[0]
    c_index=np.argmin(abs(logistic_model.Cs_-selected_c))
    lambda_rows.append([model_name,1/selected_c,logistic_model.scores_[1].mean(axis=0)[c_index]])

lambda_results=pd.DataFrame(lambda_rows,columns=["Model","$\lambda$","CV AUC"])

display(lambda_results.round(4))
```

```python
import matplotlib.pyplot as plt

plot_specifications=[
    ("three_point_attempts_per_game","Three-point attempts /gm",False),
    ("three_points_made_per_game","Three-pointers made /gm",False),
    ("two_point_attempts_per_game","Two-point attempts /gm",False),
    ("rebounds_per_game","Rebounds /gm",False),
    ("steals_per_game","Steals /gm",False),
    ("avg_age","Average age (yrs)",False),
    ("turnovers_per_game","Turnovers /gm",False),
    ("free_throws_made_per_game","Free-throws made /gm",False),
    ("avg_usage_pct","Usage %",True)
]

non_playoff_color="#c0392b"
playoff_color="#1f6fb4"

fig,axes=plt.subplots(3,3,figsize=(15,10.5))

for ax,(feature,label,is_percentage) in zip(axes.ravel(),plot_specifications):
    plot_data=team[["year",target_variable,feature]].replace([np.inf,-np.inf],np.nan).dropna()

    if is_percentage:
        plot_data[feature]*=100

    mean_values=plot_data.groupby(["year",target_variable])[feature].mean().unstack()
    std_values=plot_data.groupby(["year",target_variable])[feature].std().unstack()

    for class_label,line_color,group_name in [(0,non_playoff_color,"Non-playoff"),(1,playoff_color,"Playoff")]:
        ax.plot(mean_values.index,mean_values[class_label],color=line_color,lw=1.8,label=group_name)
        ax.fill_between(mean_values.index,mean_values[class_label]-std_values[class_label],mean_values[class_label]+std_values[class_label],color=line_color,alpha=0.12,lw=0)

    ax.set_title(label,fontsize=11,fontweight="bold",pad=10)
    ax.grid(alpha=0.25)
    ax.tick_params(labelsize=8.5)

    for spine in ("top","right"):
        ax.spines[spine].set_visible(False)

axes[0,0].legend(fontsize=8,frameon=False)
fig.tight_layout(rect=[0,0.018,1,0.955])

output_dir=Path.home()/"Desktop"/"NBA_Project"/"code"/"analysis"/"Classification"
output_dir.mkdir(parents=True,exist_ok=True)

save_path=output_dir/"Offensive Predictors Playoff Qualification.pdf"
fig.savefig(save_path,dpi=300,bbox_inches="tight")

plt.show()

print(f"Saved to: {save_path}")
```

```python
######################### Random Forest ###########################################################################
```

```python
import warnings
warnings.filterwarnings("ignore",category=UserWarning)

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold,cross_val_score

cv_rf=StratifiedKFold(n_splits=3,shuffle=True,random_state=42)
max_features_grid=[0.3,0.5,0.7,1.0]

rf_results=[]

for max_feature_value in max_features_grid:
    rf_model=RandomForestClassifier(
        max_features=max_feature_value,
        random_state=42,
        n_jobs=2
    )

    cv_scores=cross_val_score(
        rf_model,
        X_tr,
        y_tr,
        scoring="roc_auc",
        cv=cv_rf,
        n_jobs=1
    )

    rf_results.append((max_feature_value,cv_scores.mean(),cv_scores.std()))

mean_auc=np.array([result[1] for result in rf_results])
std_auc=np.array([result[2] for result in rf_results])

best_index=int(np.argmax(mean_auc))
selected_max_features=rf_results[best_index][0]

print("Selected max_features:",selected_max_features)
print("CV AUC mean:",round(mean_auc[best_index],3))
print("CV AUC SD:",round(std_auc[best_index],3))

fig,ax=plt.subplots(figsize=(8,5))

x_values=np.arange(len(rf_results))
x_labels=[str(result[0]) for result in rf_results]

ax.fill_between(x_values,mean_auc-std_auc,mean_auc+std_auc,color="#4c72b0",alpha=0.15,label="±1 SD")
ax.plot(x_values,mean_auc,"o-",color="#4c72b0",lw=2,label="CV AUC mean")
ax.plot(x_values[best_index],mean_auc[best_index],"o",color="#c44e52",markersize=9)

ax.annotate(
    f"CV optimum\nAUC = {mean_auc[best_index]:.3f}",
    xy=(x_values[best_index],mean_auc[best_index]),
    xytext=(x_values[best_index]+0.55,mean_auc[best_index]+0.013),
    fontsize=9,
    arrowprops=dict(arrowstyle="->",color="#888888"),
    bbox=dict(boxstyle="round,pad=0.35",facecolor="white",edgecolor="#c44e52")
)

ax.set_xticks(x_values)
ax.set_xticklabels(x_labels)
ax.set_xlabel("max_features")
ax.set_ylabel("3-fold cross-validated AUC")
ax.grid(axis="y",alpha=0.3)

for spine in ("top","right"):
    ax.spines[spine].set_visible(False)

ax.legend(fontsize=8.5,loc="lower right")
fig.tight_layout()

output_dir=Path.home()/"Desktop"/"NBA_Project"/"code"/"analysis"/"Classification"
output_dir.mkdir(parents=True,exist_ok=True)

save_path=output_dir/"MTRY Tuning Class.pdf"
fig.savefig(save_path,dpi=300,bbox_inches="tight")

plt.show()

selected_rf={
    "max_features":selected_max_features,
    "random_state":42,
    "n_jobs":2
}

print("Selected Random Forest settings:")
print(selected_rf)
```

```python
from sklearn.inspection import permutation_importance
from sklearn.metrics import roc_auc_score,accuracy_score,confusion_matrix
from sklearn.ensemble import RandomForestClassifier

rf_model=RandomForestClassifier(**selected_rf).fit(X_tr,y_tr)

rf_probabilities,rf_predictions=rf_model.predict_proba(X_te)[:,1],rf_model.predict(X_te)

cv_scores=cross_val_score(
    RandomForestClassifier(**selected_rf),
    X_tr,
    y_tr,
    scoring="roc_auc",
    cv=cv_rf
)

display(pd.DataFrame({
    "CV AUC mean":[cv_scores.mean()],
    "CV AUC sd":[cv_scores.std()],
    "test AUC":[roc_auc_score(y_te,rf_probabilities)],
    "test accuracy":[accuracy_score(y_te,rf_predictions)]
},index=["Random Forest"]).round(4))

display(pd.DataFrame(
    confusion_matrix(y_te,rf_predictions),
    index=["actual: no playoffs","actual: playoffs"],
    columns=["predicted: no playoffs","predicted: playoffs"]
))

permutation_results=permutation_importance(
    rf_model,
    X_te,
    y_te,
    n_repeats=10,
    random_state=42,
    scoring="roc_auc"
)

feature_importance=pd.DataFrame({
    "Permutation importance":permutation_results.importances_mean
},index=X_tr.columns).sort_values(
    "Permutation importance",
    ascending=False
)

display(feature_importance.round(3))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

fitted_all = dict(trained_models)
fitted_all["Random Forest"] = rf_model

fig, ax = plt.subplots(figsize=(8, 6.5))
fig.patch.set_facecolor("white")
ax.set_facecolor("#fafafa")

colors = ["#4c72b0", "#55a868","#dd8452", "#c44e52"]
line_styles = ["--", "-", "-.", "-"]

for (name, model), color, ls in zip(fitted_all.items(), colors, line_styles):
    prob = model.predict_proba(X_te)[:, 1]
    fpr, tpr, _ = roc_curve(y_te, prob)
    auc = roc_auc_score(y_te, prob)

    ax.plot(fpr, tpr, lw=2, linestyle=ls, color=color,
            label=f"{name} (AUC {auc:.4f})", zorder=3)

ax.plot([0, 1], [0, 1], "--", color="#aaaaaa", lw=1.2,
        label="Chance (AUC 0.5000)")

ax.set_xlabel("False positive rate", fontsize=11)
ax.set_ylabel("True positive rate", fontsize=11)
ax.tick_params(axis="both", labelsize=9)
ax.set_xlim(-0.01, 1.01)
ax.set_ylim(-0.01, 1.01)
ax.grid(color="#d9d9d9", linewidth=0.6, alpha=0.6)
ax.set_axisbelow(True)

for s in ("top", "right"):
    ax.spines[s].set_visible(False)

ax.spines["left"].set_color("#888888")
ax.spines["bottom"].set_color("#888888")
ax.legend(fontsize=9, loc="lower right", frameon=True)

fig.tight_layout()

output_dir = Path.home() / "Desktop" / "NBA_Project" / "code" / "analysis" / "Classification"
output_dir.mkdir(parents=True, exist_ok=True)

save_path = output_dir / "ROC Curves.pdf"
fig.savefig(save_path, dpi=300, bbox_inches="tight")

plt.show()
```
