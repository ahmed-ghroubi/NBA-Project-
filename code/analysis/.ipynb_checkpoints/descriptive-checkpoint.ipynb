```python
############################################# box plot: fantasy points by weight ranges #############################################

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create clean plotting dataset for weight analysis
weight_box_data = nba_final[
    [
        "weight_kg",
        "nba_fantasy_points"
    ]
].copy()

# Make sure columns are numeric
weight_box_data["weight_kg"] = pd.to_numeric(
    weight_box_data["weight_kg"],
    errors="coerce"
)

weight_box_data["nba_fantasy_points"] = pd.to_numeric(
    weight_box_data["nba_fantasy_points"],
    errors="coerce"
)

# Remove missing values
weight_box_data = weight_box_data.dropna()

############################################# create 5 kg weight groups #############################################

# Create 5 kg weight groups
weight_min = int(np.floor(weight_box_data["weight_kg"].min() / 5) * 5)
weight_max = int(np.ceil(weight_box_data["weight_kg"].max() / 5) * 5)

weight_bins = list(range(weight_min, weight_max + 5, 5))

weight_box_data["weight_group_5kg"] = pd.cut(
    weight_box_data["weight_kg"],
    bins=weight_bins,
    right=False
)

############################################# prepare box plot data #############################################

weight_groups = weight_box_data["weight_group_5kg"].cat.categories

weight_boxplot_data = []
weight_labels = []

for group in weight_groups:
    group_values = weight_box_data[
        weight_box_data["weight_group_5kg"] == group
    ]["nba_fantasy_points"]
    
    # Only keep groups that actually contain players
    if len(group_values) > 0:
        weight_boxplot_data.append(group_values)
        weight_labels.append(f"{int(group.left)}-{int(group.right)}")

############################################# create box plot #############################################

plt.figure(figsize=(14, 5))

plt.boxplot(
    weight_boxplot_data,
    labels=weight_labels
)

plt.title("NBA Fantasy Points Distribution by Weight Group")
plt.xlabel("Weight group in kg")
plt.ylabel("NBA Fantasy Points")
plt.xticks(rotation=45)
plt.show()
```

```python
############################################# box plot: fantasy points by exact age #############################################

import matplotlib.pyplot as plt
import pandas as pd

# Use nba_fantasy_points as overall player performance measure
performance_variable = "nba_fantasy_points"

# Create a clean plotting dataset
age_plot_data = nba_final[
    [
        "age",
        performance_variable
    ]
].dropna().copy()

# Make sure age is numeric
age_plot_data["age"] = pd.to_numeric(
    age_plot_data["age"],
    errors="coerce"
)

# Keep only realistic NBA ages
age_plot_data = age_plot_data[
    (age_plot_data["age"] >= 18) &
    (age_plot_data["age"] <= 45)
].copy()

# Sort ages
ages = sorted(age_plot_data["age"].unique())

# Create one box for each age
boxplot_data = [
    age_plot_data[age_plot_data["age"] == age][performance_variable]
    for age in ages
]

plt.figure(figsize=(14, 5))
plt.boxplot(
    boxplot_data,
    labels=ages
)

plt.title("NBA Fantasy Points Distribution by Exact Age")
plt.xlabel("Age")
plt.ylabel("NBA Fantasy Points")
plt.xticks(rotation=45)
plt.show()
```

```python
############################################# box plot: fantasy points by each year of player experience #############################################

import pandas as pd
import matplotlib.pyplot as plt

# Create clean plotting dataset
experience_box_data = nba_final[
    [
        "player_experience_years",
        "nba_fantasy_points"
    ]
].dropna().copy()

# Make sure columns are numeric
experience_box_data["player_experience_years"] = pd.to_numeric(
    experience_box_data["player_experience_years"],
    errors="coerce"
)

experience_box_data["nba_fantasy_points"] = pd.to_numeric(
    experience_box_data["nba_fantasy_points"],
    errors="coerce"
)

# Sort experience years
experience_years = sorted(
    experience_box_data["player_experience_years"].unique()
)

# Create one box for each experience year
experience_boxplot_data = [
    experience_box_data[
        experience_box_data["player_experience_years"] == experience
    ]["nba_fantasy_points"]
    for experience in experience_years
]

# Create labels
experience_labels = [
    str(int(experience))
    for experience in experience_years
]

plt.figure(figsize=(14, 5))

plt.boxplot(
    experience_boxplot_data,
    labels=experience_labels
)

plt.title("NBA Fantasy Points Distribution by Each Year of Player Experience")
plt.xlabel("Player Experience in Years")
plt.ylabel("NBA Fantasy Points")
plt.xticks(rotation=45)
plt.show()
```
