```python
##################### Filtering variables #######################################################################
```

```python
# ============================ STEP 1 ==================================

# Candidate predictors and their correlation with the per-game target.

import numpy as np, pandas as pd, matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv("/Users/eyaghroubi/Desktop/nba_final_player_season_data.csv")
print(f"Original rows: {df.shape[0]:,}   Columns: {df.shape[1]}")


# Create per-game target and per-game component variables.
df = df.replace([np.inf, -np.inf], np.nan)
df = df[df["games_played"] > 0].copy()

BASE_COMPONENTS = ["points", "rebounds", "assists", "steals", "blocks", "turnovers"]

VOLUME_VARS = BASE_COMPONENTS + [
    "plus_minus",

    "offensive_rebounds", "defensive_rebounds",
    "blocks_against", "personal_fouls", "personal_fouls_drawn",

    "two_point_points", "three_point_points",

    "two_point_attempts", "three_point_attempts", "free_throw_attempts",
    "field_goal_attempts", "field_goals_made",
    "two_points_made", "three_points_made", "free_throws_made"
]
VOLUME_VARS = [c for c in VOLUME_VARS if c in df.columns]

for c in VOLUME_VARS:
    df[f"{c}_per_game"] = df[c] / df["games_played"]

df["nba_fantasy_points_per_game"] = df["nba_fantasy_points"] / df["games_played"]

# Remove players with very few games because per-game values are unstable.

MIN_GAMES = 20
df = df[df["games_played"] >= MIN_GAMES].copy()

print(f"Rows after minimum-games filter: {df.shape[0]:,}")
print(f"Minimum games played: {MIN_GAMES}")

TARGET = "nba_fantasy_points_per_game"

COMPONENTS = [
    "points_per_game", "rebounds_per_game", "assists_per_game",
    "steals_per_game", "blocks_per_game", "turnovers_per_game"
]

# Check whether the per-game fantasy-points formula is correct.

fp_check = df[TARGET] - (
    df["points_per_game"]
    + 1.2 * df["rebounds_per_game"]
    + 1.5 * df["assists_per_game"]
    + 3 * df["steals_per_game"]
    + 3 * df["blocks_per_game"]
    - df["turnovers_per_game"]
)

print(f"max formula difference: {fp_check.abs().max():.3e}")

EXCLUDE = BASE_COMPONENTS + COMPONENTS + [TARGET, "nba_fantasy_points"] + VOLUME_VARS + [
    "games_played", "minutes",
    "player_id", "team_id", "year", "career_start_year", "career_end_year",
    "draft_year", "roster_status",
    "actual_playoff_team", "top_8_conference_team", "playoff_seed",
    "playoff_finish_rank", "reached_conference_semifinals",
    "reached_conference_finals", "reached_finals", "won_nba_title",
]

CANDIDATES = [
    c for c in df.select_dtypes("number")
    if c not in EXCLUDE and df[c].std() > 0
]

corr = pd.DataFrame({
    "with_target":    df[CANDIDATES].corrwith(df[TARGET]).abs(),
    "with_component": df[CANDIDATES].apply(lambda s: df[COMPONENTS].corrwith(s).abs().max()),
}).round(3)

corr["criterion"] = corr.max(axis=1)
corr = corr.sort_values("criterion", ascending=False)

display(corr)
print(f"\nCandidate predictors after initial exclusion: {len(CANDIDATES)}")

for c in CANDIDATES:
    print("  ", c)
```

```python
# ============================ STEP 2 ==================================
# Remove predictors that are too close to the target or its components.


# Variables computed from the components of the target. They are rescaled
# versions of the outcome and are removed regardless of their correlation.
BY_DEFINITION = [
    "triple_doubles",           # 10+ in three of PTS / REB / AST
    "double_doubles",           # 10+ in two of PTS / REB / AST
    "usage_pct",                # from shot attempts, free throws, turnovers
    "assist_pct",               # from assists
    "offensive_rebound_pct",    # from rebounds
    "defensive_rebound_pct",    # from rebounds
]

survivors = [c for c in corr.index[corr.criterion <= 0.55] if c not in BY_DEFINITION]
display(corr.loc[survivors])
```

```python
from matplotlib.patches import Patch

TH = 0.55 # Threshold
p = corr.sort_values("criterion", ascending=False)

col = ["#dd8452" if n in BY_DEFINITION else ("#c44e52" if v > TH else "#4c72b0")
       for n, v in p["criterion"].items()]

fig, ax = plt.subplots(figsize=(11, 4.8))
ax.bar(p.index, p["criterion"], color=col)
ax.axhline(TH, color="#333333", ls="--", lw=1.2)
ax.text(len(p) - 0.5, TH + 0.02, f"threshold {TH}", ha="right", fontsize=8)
ax.set_ylabel("max |corr| with target\nor component", fontsize=9)
plt.setp(ax.get_xticklabels(), rotation=90, fontsize=7.5)
ax.set_xlim(-0.7, len(p) - 0.3)
#ax.set_title("Predictor screening", fontsize=11)
ax.legend(handles=[
    Patch(color="#4c72b0", label="retained"),
    Patch(color="#dd8452", label="removed: computed from components"),
    Patch(color="#c44e52", label="removed: correlation > 0.55"),
], fontsize=8, loc="upper right")
fig.tight_layout()
plt.show()
```

```python
# ============================ STEP 3 ==================================
# Redundancy between predictors: VIF.
# Step 2 removed variables close to the TARGET.
# This step removes variables that duplicate EACH OTHER.

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

def vif_table(cols):
    X = sm.add_constant(df[cols].dropna())
    return (pd.Series([variance_inflation_factor(X.values, i + 1) for i in range(len(cols))],
                      index=cols, name="VIF")
            .sort_values(ascending=False))

display(vif_table(survivors).round(1))

# Redundant groups, removed with a stated reason rather than automatically:
#   team strength : diff = points - opponent points, wins + losses = games
#                   -> keep team_point_diff_per_game only
#   shooting      : TS% already combines 2P, 3P and FT efficiency
DROP = [
    "team_points_per_game", "team_opp_points_per_game",
    "team_wins", "team_losses", "team_win_pct",
    "wins", "losses", "win_pct",
    "field_goal_pct", "two_point_pct", "three_point_pct", "free_throw_pct"
]
final_vars = [c for c in survivors if c not in DROP]

print(f"{len(survivors)} -> {len(final_vars)} predictors\n")
display(vif_table(final_vars).round(2))
```

```python
########################################### OLS ###################################################################
```

```python
# Train/test split and baseline linear model.
#
# The split is by PLAYER, not by row: a player appears in several seasons,
# so splitting rows would put the same player in both sets and the test
# set would no longer be unseen.

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

data = df[final_vars + [TARGET, "player_id"]].dropna()

tr_idx, te_idx = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
                      .split(data, groups=data["player_id"]))
train, test = data.iloc[tr_idx], data.iloc[te_idx]

X_tr, y_tr = train[final_vars], train[TARGET]
X_te, y_te = test[final_vars],  test[TARGET]

print(f"train {len(train):,} rows / {train.player_id.nunique()} players")
print(f"test  {len(test):,} rows / {test.player_id.nunique()} players")
print(f"overlap: {len(set(train.player_id) & set(test.player_id))} players")

# --- OLS with heteroscedasticity-robust (HC3) standard errors ---------
ols  = sm.OLS(y_tr, sm.add_constant(X_tr)).fit(cov_type="HC3")
pred = ols.predict(sm.add_constant(X_te))

# --- performance -------------------------------------------------------
perf = pd.Series({
    "adj. R2 (in-sample)": ols.rsquared_adj,
    "R2 (test)":           r2_score(y_te, pred),
    "MAE (test)":          mean_absolute_error(y_te, pred),
    "RMSE (test)":         np.sqrt(mean_squared_error(y_te, pred)),
}, name="OLS").round(3)
display(perf)

# --- coefficients ------------------------------------------------------
# Raw coefficients are on the original scale and cannot be compared with
# each other. The standardized column gives the change in fantasy points
# per one standard deviation of the predictor, which is comparable.
scaler  = StandardScaler().fit(X_tr)
ols_std = sm.OLS(y_tr, sm.add_constant(scaler.transform(X_tr))).fit(cov_type="HC3")

coefs = pd.DataFrame({
    "coefficient":  ols.params.drop("const"),
    "std. error":   ols.bse.drop("const"),
    "p-value":      ols.pvalues.drop("const"),
    "CI lower":     ols.conf_int().drop("const")[0],
    "CI upper":     ols.conf_int().drop("const")[1],
    "standardized": ols_std.params[1:].values,
})
display(coefs.reindex(coefs["standardized"].abs().sort_values(ascending=False).index).round(3))
```

```python
########################### Linearity check ###################################################################
```

```python
from statsmodels.stats.diagnostic import linear_reset
from statsmodels.nonparametric.smoothers_lowess import lowess

resid = ols.resid

fig, axes = plt.subplots(3, 3, figsize=(13, 10))
for ax, v in zip(axes.ravel(), final_vars):
    x  = X_tr[v].to_numpy()
    pr = (resid + ols.params[v] * X_tr[v]).to_numpy()

    ax.scatter(x, pr, s=2, alpha=0.06, color="grey")
    smooth = lowess(pr, x, frac=0.4, return_sorted=True)
    ax.plot(smooth[:, 0], smooth[:, 1], color="crimson", lw=2)

    xs = np.linspace(x.min(), x.max(), 50)
    ax.plot(xs, ols.params[v] * xs, "--", color="navy", lw=1.3)
    ax.set_title(v, fontsize=10)

fig.suptitle("Partial residual plots: LOWESS (red) vs. linear effect (blue)")
fig.tight_layout()
plt.show()
```

```python
########################## Polynomial #######################################################################
```

```python
# Variables showing clear nonlinear patterns in the LOWESS plots
NONLINEAR = [
    "plus_minus_per_game",
    "player_experience_years",
    "true_shooting_pct",
    "age",
    "height_cm",
    "net_rating"
]

print("Nonlinear terms added for:", NONLINEAR)

# Standardize using training-set mean and standard deviation
mu, sd = X_tr.mean(), X_tr.std()

def build(X):
    Z = (X - mu) / sd
    out = Z.copy()

    for v in NONLINEAR:
        out[f"{v}_sq"] = Z[v] ** 2
        out[f"{v}_cu"] = Z[v] ** 3

    return out

P_tr = build(X_tr)
P_te = build(X_te)

# Polynomial OLS model
poly = sm.OLS(
    y_tr,
    sm.add_constant(P_tr)
).fit(cov_type="HC3")

pred_poly = poly.predict(
    sm.add_constant(P_te)
)

# Compare linear and polynomial models
comparison = pd.DataFrame({
    "predictors": [
        len(final_vars),
        P_tr.shape[1]
    ],
    "adj. R2 (in-sample)": [
        ols.rsquared_adj,
        poly.rsquared_adj
    ],
    "R2 (test)": [
        r2_score(y_te, pred),
        r2_score(y_te, pred_poly)
    ],
    "MAE (test)": [
        mean_absolute_error(y_te, pred),
        mean_absolute_error(y_te, pred_poly)
    ],
    "RMSE (test)": [
        np.sqrt(mean_squared_error(y_te, pred)),
        np.sqrt(mean_squared_error(y_te, pred_poly))
    ],
}, index=["OLS linear", "OLS polynomial"]).rename_axis(None)

display(comparison.round(3))

# Polynomial coefficients
coefs_poly = pd.DataFrame({
    "coefficient": poly.params.drop("const"),
    "std. error": poly.bse.drop("const"),
    "p-value": poly.pvalues.drop("const"),
    "CI lower": poly.conf_int().drop("const")[0],
    "CI upper": poly.conf_int().drop("const")[1],
}).rename_axis(None)

display(coefs_poly.round(3))
```

```python
########################## GAM #######################################################################
```

```python
from pygam import LinearGAM, s, l

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

SMOOTH = NONLINEAR
LINEAR = [v for v in final_vars if v not in SMOOTH]

terms = None

for i, v in enumerate(final_vars):
    term = s(i) if v in SMOOTH else l(i)
    terms = term if terms is None else terms + term

gam = LinearGAM(terms).gridsearch(
    X_tr[final_vars].values,
    np.asarray(y_tr).ravel(),
    progress=False
)


pred_gam = gam.predict(X_te[final_vars].values)

comparison = pd.DataFrame({
    "R2 (test)": [
        r2_score(y_te, pred),
        r2_score(y_te, pred_poly),
        r2_score(y_te, pred_gam)
    ],
    "MAE (test)": [
        mean_absolute_error(y_te, pred),
        mean_absolute_error(y_te, pred_poly),
        mean_absolute_error(y_te, pred_gam)
    ],
    "RMSE (test)": [
        np.sqrt(mean_squared_error(y_te, pred)),
        np.sqrt(mean_squared_error(y_te, pred_poly)),
        np.sqrt(mean_squared_error(y_te, pred_gam))
    ],
}, index=[
    "OLS linear",
    "OLS polynomial",
    "GAM"
]).rename_axis(None)

display(comparison.round(3))
print(f"effective degrees of freedom: {gam.statistics_['edof']:.3f}")
```

```python
from sklearn.model_selection import GroupKFold
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score

groups_tr = train["player_id"]
gkf = GroupKFold(n_splits=5)

ns = np.array(sorted(set(np.round(np.logspace(0, 3, 30)).astype(int)) | {500}))

fold_scores = []
for f_tr, f_va in gkf.split(X_tr, groups=groups_tr):
    A_X, A_y = X_tr.iloc[f_tr], y_tr.iloc[f_tr]
    V_X, V_y = X_tr.iloc[f_va], y_tr.iloc[f_va]

    m = RandomForestRegressor(n_estimators=1000, max_depth=None, min_samples_leaf=3,
                              max_features=0.7, random_state=42, n_jobs=-1).fit(A_X, A_y)

    P = np.column_stack([t.predict(V_X.values) for t in m.estimators_])
    fold_scores.append([r2_score(V_y, P[:, :n].mean(axis=1)) for n in ns])

fold_scores = np.array(fold_scores)
mu, sd = fold_scores.mean(0), fold_scores.std(0)

# One-standard-error rule.

n_folds = fold_scores.shape[0]

best_idx = int(np.argmax(mu))
best_n = int(ns[best_idx])
best_r2 = float(mu[best_idx])

se = sd / np.sqrt(n_folds)
threshold = best_r2 - se[best_idx]

plateau_idx = int(np.where(mu >= threshold)[0][0])
plateau_n = int(ns[plateau_idx])
plateau_r2 = float(mu[plateau_idx])

# Final selected value.

selected_n_estimators = 500
y_sel = float(mu[ns == selected_n_estimators][0])
conv = float(mu[-1])

print(f"Best CV R2: {best_r2:.4f} at {best_n} trees")
print(f"SD at best point: {sd[best_idx]:.4f}")
print(f"SE at best point: {se[best_idx]:.4f}")
print(f"One-SE threshold = best CV R2 - SE = {best_r2:.4f} - {se[best_idx]:.4f} = {threshold:.4f}")
print(f"One-SE point: {plateau_n} trees, CV R2 = {plateau_r2:.4f}")
print(f"selected_n_estimators = {selected_n_estimators}")

fig, ax = plt.subplots(figsize=(8.5, 4.6))

ax.fill_between(ns, mu - sd, mu + sd, color="steelblue", alpha=0.15, lw=0,
                label="$\\pm$1 SD across folds")
ax.plot(ns, mu, color="steelblue", lw=2, label="5-fold CV mean")
ax.axhline(conv, color="grey", ls=":", lw=1)

ax.plot([plateau_n], [plateau_r2], "o", color="dimgray", ms=7, zorder=5)
ax.plot([selected_n_estimators], [y_sel], "o", color="crimson", ms=8, zorder=6)

ax.annotate(f"one-SE point\n{plateau_n} trees",
            xy=(plateau_n, plateau_r2),
            xytext=(20, -65),
            textcoords="offset points",
            fontsize=9, color="dimgray", ha="center", va="top",
            arrowprops=dict(arrowstyle="->", color="dimgray", lw=1.2,
                            shrinkB=8))

ax.annotate(f"selected: {selected_n_estimators} trees\n$R^2$ = {y_sel:.3f}",
            xy=(selected_n_estimators, y_sel),
            xytext=(0, -70),
            textcoords="offset points",
            fontsize=10, color="crimson", ha="center", va="top",
            arrowprops=dict(arrowstyle="->", color="crimson", lw=1.4,
                            shrinkB=10))
ax.set_xscale("log")
ax.set_xlabel("number of trees (log scale)")
ax.set_ylabel("cross-validated $R^2$")
ax.set_title("Random Forest: cross-validated performance vs. number of trees",
             fontsize=11)

ax.set_ylim(0.655, 0.74)
ax.grid(alpha=0.25, which="both")
ax.legend(loc="lower right", fontsize=8.5)

for s in ("top", "right"):
    ax.spines[s].set_visible(False)

fig.tight_layout()
plt.show()
```

```python
BASE = dict(n_estimators=500, max_depth=None, min_samples_leaf=5,
            max_features=0.7, random_state=42, n_jobs=-1)
SELECTED = {"max_depth": None, "min_samples_leaf": 3, "max_features": 0.7}
GRIDS = {"max_depth":        [4, 6, 8, 10, 12, 16, None],
         "min_samples_leaf": [1, 2, 3, 5, 10, 20, 50],
         "max_features":     [0.2, 0.33, 0.5, 0.7, 1.0]}

curves = {}
for param, values in GRIDS.items():
    rows = []
    for v in values:
        p = dict(BASE); p[param] = v
        s = cross_val_score(RandomForestRegressor(**p), X_tr, y_tr, scoring="r2",
                            cv=GroupKFold(n_splits=5).split(X_tr, groups=groups_tr))
        rows.append((v, s.mean(), s.std()))
    curves[param] = rows

fig, axes = plt.subplots(1, 3, figsize=(13, 4.3))
for ax, (param, rows) in zip(axes, curves.items()):
    labels = ["None" if r[0] is None else str(r[0]) for r in rows]
    mu = np.array([r[1] for r in rows])
    sd = np.array([r[2] for r in rows])
    xs = np.arange(len(rows))

    ax.fill_between(xs, mu - sd, mu + sd, color="steelblue", alpha=0.15, lw=0)
    ax.plot(xs, mu, "o-", color="steelblue", lw=1.9, ms=4.5)

    j    = labels.index("None" if SELECTED[param] is None else str(SELECTED[param]))
    best = int(np.argmax(mu))
    lo, hi = (mu - sd).min(), (mu + sd).max()
    span = hi - lo
    ax.set_ylim(lo - 0.28 * span, hi + 0.10 * span)

    ax.plot([xs[j]], [mu[j]], "o", color="crimson", ms=4.5, zorder=5)
    ax.annotate("selected", xy=(xs[j], mu[j]-0.002), xytext=(xs[j], mu[j] - 0.19 * span),
                ha="center", fontsize=9, color="crimson",
                arrowprops=dict(arrowstyle="->", color="crimson", lw=1.2))

    if best != j:
        ax.plot([xs[best]], [mu[best]], "o", color="darkorange", ms=4.5, zorder=5)
        ax.annotate("CV optimum", xy=(xs[best], mu[best]-0.002),
                    xytext=(xs[best], mu[best] - 0.19 * span),
                    ha="center", fontsize=9, color="darkorange",
                    arrowprops=dict(arrowstyle="->", color="darkorange", lw=1.2))

    ax.set_xticks(xs)
    ax.set_xticklabels(labels)
    ax.set_xlabel(param)
    ax.grid(alpha=0.25)
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)

axes[0].set_ylabel("5-fold cross-validated $R^2$")
#fig.suptitle("Random Forest: validation curves (one parameter varied, others held fixed)",
 #            fontsize=11)
fig.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance

rf = RandomForestRegressor(n_estimators=500, max_depth=None, min_samples_leaf=3,
                           max_features=0.7, random_state=42, n_jobs=-1).fit(X_tr, y_tr)
pred_rf = rf.predict(X_te)

comparison = pd.DataFrame({
    "R2 (test)":   [r2_score(y_te, pred), r2_score(y_te, pred_poly),
                    r2_score(y_te, pred_gam), r2_score(y_te, pred_rf)],
    "MAE (test)":  [mean_absolute_error(y_te, pred),
                    mean_absolute_error(y_te, pred_poly),
                    mean_absolute_error(y_te, pred_gam),
                    mean_absolute_error(y_te, pred_rf)],
    "RMSE (test)": [np.sqrt(mean_squared_error(y_te, pred)),
                    np.sqrt(mean_squared_error(y_te, pred_poly)),
                    np.sqrt(mean_squared_error(y_te, pred_gam)),
                    np.sqrt(mean_squared_error(y_te, pred_rf))],
}, index=["OLS linear", "OLS polynomial", "GAM",
          "Random Forest"]).rename_axis(None)
display(comparison.round(3))

perm = permutation_importance(rf, X_te, y_te, n_repeats=10,
                              random_state=42, n_jobs=-1)

importance = pd.DataFrame({
    "std. coefficient": np.abs(ols_std.params[1:].values),
    "RF impurity":      rf.feature_importances_,
    "permutation":      perm.importances_mean,
}, index=final_vars)
for c in list(importance.columns):
    importance[c + " rank"] = importance[c].rank(ascending=False).astype(int)
display(importance.round(3))

```
