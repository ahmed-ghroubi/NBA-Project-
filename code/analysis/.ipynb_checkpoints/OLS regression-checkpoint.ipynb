```python
############################################# load final dataset #############################################

import pandas as pd
import matplotlib.pyplot as plt

# Load final player-season dataset
nba_final = pd.read_csv(
   r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\nba_final_player_season_data.csv"
)

print("Dataset shape:", nba_final.shape)

nba_final.head()
```

```python
############################################# ANOVA: fantasy points by experience group #############################################
```

```python
import pandas as pd
import scipy.stats as stats

anova_data = nba_final[
    [
        "player_experience_years",
        "nba_fantasy_points"
    ]
].copy()

anova_data["player_experience_years"] = pd.to_numeric(
    anova_data["player_experience_years"],
    errors="coerce"
)

anova_data["nba_fantasy_points"] = pd.to_numeric(
    anova_data["nba_fantasy_points"],
    errors="coerce"
)

anova_data = anova_data.dropna()

anova_data["experience_group"] = pd.cut(
    anova_data["player_experience_years"],
    bins=[-1, 2, 5, 10, 15, 40],
    labels=[
        "0-2 years",
        "3-5 years",
        "6-10 years",
        "11-15 years",
        "16+ years"
    ]
)

groups = [
    group["nba_fantasy_points"].values
    for name, group in anova_data.groupby("experience_group")
]

f_statistic, p_value = stats.f_oneway(*groups)

print("ANOVA result")
print("F-statistic:", f_statistic)
print("p-value:", p_value)
```

```python
############################################# Tukey HSD test: fantasy points by experience group #############################################

from statsmodels.stats.multicomp import pairwise_tukeyhsd
import pandas as pd

# Remove missing experience groups if any exist
tukey_data = anova_data.dropna(
    subset=[
        "experience_group",
        "nba_fantasy_points"
    ]
).copy()

# Tukey HSD test
tukey_result = pairwise_tukeyhsd(
    endog=tukey_data["nba_fantasy_points"],
    groups=tukey_data["experience_group"],
    alpha=0.05
)

print(tukey_result)
```

```python
############################################# Tukey HSD result as table #############################################

tukey_table = pd.DataFrame(
    data=tukey_result._results_table.data[1:],
    columns=tukey_result._results_table.data[0]
)

display(tukey_table)
```

```python
############################################# show only significant Tukey comparisons #############################################

significant_tukey_results = tukey_table[
    tukey_table["reject"] == True
].copy()

display(significant_tukey_results)
```

```python
############################################# mean fantasy points by experience group #############################################

experience_group_summary = tukey_data.groupby("experience_group")["nba_fantasy_points"].agg(
    ["count", "mean", "median", "std"]
)

display(experience_group_summary)
```

```python
############################################# linear regression: explaining fantasy points #############################################
```

```python
import pandas as pd
import statsmodels.api as sm

regression_cols = [
    "nba_fantasy_points",
    "minutes",
    "age",
    "height_cm",
    "weight_kg",
    "player_experience_years",
    "usage_pct",
    "true_shooting_pct",
    "plus_minus"
]

regression_data = nba_final[regression_cols].copy()

for col in regression_cols:
    regression_data[col] = pd.to_numeric(regression_data[col], errors="coerce")

regression_data = regression_data.dropna()

X = regression_data[
    [
        "minutes",
        "age",
        "height_cm",
        "weight_kg",
        "player_experience_years",
        "usage_pct",
        "true_shooting_pct",
        "plus_minus"
    ]
]

y = regression_data["nba_fantasy_points"]

# Add intercept
X = sm.add_constant(X)

linear_model = sm.OLS(y, X).fit()

print(linear_model.summary())
```
