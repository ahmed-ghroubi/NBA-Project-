```python
############################################# load raw player performance stats #################

import pandas as pd
import time
from nba_api.stats.endpoints import LeagueDashPlayerStats

all_seasons = []

def make_season_string(end_year):
    """
    Example:
    1998 -> '1997-98'
    1999 -> '1998-99'
    2000 -> '1999-00'
    2025 -> '2024-25'
    """
    return f"{end_year - 1}-{str(end_year)[-2:]}"

# Load NBA player performance data from 1997-98 to 2025-26
for end_year in range(1998, 2027):
    season = make_season_string(end_year)
    print("Loading performance stats:", season)
    
    try:
        stats = LeagueDashPlayerStats(
            season=season,
            season_type_all_star="Regular Season",
            per_mode_detailed="PerGame",
            timeout=60
        )
        
        season_df = stats.get_data_frames()[0]
        season_df["Season"] = season
        season_df["Year"] = end_year
        
        all_seasons.append(season_df)
        
        time.sleep(2)
        
    except Exception as e:
        print("Could not load:", season)
        print(e)

# Combine all seasons into one dataframe
nba_stats_official = pd.concat(all_seasons, ignore_index=True)

# Save raw performance data
nba_stats_official.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\raw\nba_player_performance_raw.csv",
    index=False
)

# Check raw data
print("Raw performance data shape:", nba_stats_official.shape)

nba_stats_official.head()
```

```python
############################################# clean player performance stats #################

# Remove rank columns because they are only leaderboard ranks, not actual performance values
rank_columns = [col for col in nba_stats_official.columns if col.endswith("_RANK")]

nba_performance = nba_stats_official.drop(columns=rank_columns).copy()

# Keep and reorder useful columns
nba_performance = nba_performance[
    [
        # Season information
        "Year",
        "Season",

        # Player identity
        "PLAYER_ID",
        "PLAYER_NAME",

        # Team information
        "TEAM_ID",
        "TEAM_ABBREVIATION",
        "TEAM_COUNT",

        # Player / playing time
        "AGE",
        "GP",
        "MIN",

        # Team result variables
        "W",
        "L",
        "W_PCT",

        # Scoring
        "PTS",
        "FGM",
        "FGA",
        "FG_PCT",
        "FG3M",
        "FG3A",
        "FG3_PCT",
        "FTM",
        "FTA",
        "FT_PCT",

        # Rebounding
        "OREB",
        "DREB",
        "REB",

        # Playmaking / ball handling
        "AST",
        "TOV",

        # Defense
        "STL",
        "BLK",
        "BLKA",

        # Fouls
        "PF",
        "PFD",

        # Impact / extra stats
        "PLUS_MINUS",
        "NBA_FANTASY_PTS",
        "DD2",
        "TD3"
    ]
].copy()

# Rename columns to cleaner names
nba_performance = nba_performance.rename(columns={
    "Year": "year",
    "Season": "season",
    "PLAYER_ID": "player_id",
    "PLAYER_NAME": "player_name",
    "TEAM_ID": "team_id",
    "TEAM_ABBREVIATION": "team",
    "TEAM_COUNT": "team_count",
    "AGE": "age",
    "GP": "games_played",
    "MIN": "minutes",
    "W": "wins",
    "L": "losses",
    "W_PCT": "win_pct",
    "PTS": "points",
    "FGM": "field_goals_made",
    "FGA": "field_goal_attempts",
    "FG_PCT": "field_goal_pct",
    "FG3M": "three_points_made",
    "FG3A": "three_point_attempts",
    "FG3_PCT": "three_point_pct",
    "FTM": "free_throws_made",
    "FTA": "free_throw_attempts",
    "FT_PCT": "free_throw_pct",
    "OREB": "offensive_rebounds",
    "DREB": "defensive_rebounds",
    "REB": "rebounds",
    "AST": "assists",
    "TOV": "turnovers",
    "STL": "steals",
    "BLK": "blocks",
    "BLKA": "blocks_against",
    "PF": "personal_fouls",
    "PFD": "personal_fouls_drawn",
    "PLUS_MINUS": "plus_minus",
    "NBA_FANTASY_PTS": "nba_fantasy_points",
    "DD2": "double_doubles",
    "TD3": "triple_doubles"
})

# Remove .0 from age if possible
nba_performance["age"] = pd.to_numeric(nba_performance["age"], errors="coerce").astype("Int64")

# Save cleaned performance data
nba_performance.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\nba_performance_clean.csv",
    index=False
)

print("Original nba_stats_official shape:", nba_stats_official.shape)
print("Removed rank columns:", len(rank_columns))
print("Clean nba_performance shape:", nba_performance.shape)

nba_performance.head(50)
```

```python
############################################# merge bio/career data with performance data #################

import pandas as pd

# Load cleaned bio + career dataset
bio_career_info = pd.read_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\bio_career_info_clean.csv"
)

# Load cleaned performance dataset
nba_performance = pd.read_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\nba_performance_clean.csv"
)

# Make sure merge keys have the same type
bio_career_info["player_id"] = pd.to_numeric(bio_career_info["player_id"], errors="coerce").astype("Int64")
bio_career_info["year"] = pd.to_numeric(bio_career_info["year"], errors="coerce").astype("Int64")

nba_performance["player_id"] = pd.to_numeric(nba_performance["player_id"], errors="coerce").astype("Int64")
nba_performance["year"] = pd.to_numeric(nba_performance["year"], errors="coerce").astype("Int64")

# Check duplicate keys before merging
merge_keys = ["player_id", "year"]

print("Duplicates in bio_career_info:", bio_career_info[merge_keys].duplicated().sum())
print("Duplicates in nba_performance:", nba_performance[merge_keys].duplicated().sum())

# Keep only extra bio/career columns to avoid duplicate columns like points_x and points_y
bio_career_extra = bio_career_info[
    [
        "player_id",
        "year",
        "career_start_year",
        "career_end_year",
        "roster_status",
        "height_cm",
        "weight_kg",
        "college",
        "country",
        "draft_year",
        "draft_round",
        "draft_number",
        "net_rating",
        "usage_pct",
        "true_shooting_pct",
        "assist_pct",
        "offensive_rebound_pct",
        "defensive_rebound_pct"
    ]
].copy()

# If there are duplicate player-year rows in bio/career, keep one row per player-season
bio_career_extra = bio_career_extra.drop_duplicates(subset=merge_keys)

# Merge performance data with bio/career information
# nba_performance is the main table
nba_final = nba_performance.merge(
    bio_career_extra,
    on=merge_keys,
    how="left",
    validate="many_to_one"
)

# Check merge result
print("nba_performance shape:", nba_performance.shape)
print("bio_career_extra shape:", bio_career_extra.shape)
print("nba_final shape:", nba_final.shape)

print("Missing career_start_year:", nba_final["career_start_year"].isna().sum())
print("Missing height_cm:", nba_final["height_cm"].isna().sum())
print("Missing draft_year:", nba_final["draft_year"].isna().sum())
```

```python
############################################# reorder and save final dataset #################

# Reorder columns nicely
preferred_order = [
    # Season information
    "year",
    "season",

    # Player identity
    "player_id",
    "player_name",

    # Team information
    "team_id",
    "team",
    "team_count",

    # Player age and career information
    "age",
    "career_start_year",
    "career_end_year",
    "roster_status",

    # Physical and background information
    "height_cm",
    "weight_kg",
    "college",
    "country",

    # Draft information
    "draft_year",
    "draft_round",
    "draft_number",

    # Playing time and team result
    "games_played",
    "minutes",
    "wins",
    "losses",
    "win_pct",

    # Scoring
    "points",
    "field_goals_made",
    "field_goal_attempts",
    "field_goal_pct",
    "three_points_made",
    "three_point_attempts",
    "three_point_pct",
    "free_throws_made",
    "free_throw_attempts",
    "free_throw_pct",

    # Rebounding
    "offensive_rebounds",
    "defensive_rebounds",
    "rebounds",

    # Playmaking / ball handling
    "assists",
    "turnovers",

    # Defense
    "steals",
    "blocks",
    "blocks_against",

    # Fouls
    "personal_fouls",
    "personal_fouls_drawn",

    # Impact and extra stats
    "plus_minus",
    "nba_fantasy_points",
    "double_doubles",
    "triple_doubles",

    # Advanced bio/performance percentages
    "net_rating",
    "usage_pct",
    "true_shooting_pct",
    "assist_pct",
    "offensive_rebound_pct",
    "defensive_rebound_pct"
]

# Keep only columns that exist, then append any remaining columns at the end
preferred_order = [col for col in preferred_order if col in nba_final.columns]
remaining_columns = [col for col in nba_final.columns if col not in preferred_order]

nba_final = nba_final[preferred_order + remaining_columns]

# Remove .0 from integer columns
for col in ["year", "player_id", "team_id", "age", "career_start_year", "career_end_year"]:
    if col in nba_final.columns:
        nba_final[col] = pd.to_numeric(nba_final[col], errors="coerce").astype("Int64")

# Save final merged dataset
nba_final.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\nba_final_player_season_data.csv",
    index=False
)

print("Saved final dataset:")
print(r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\nba_final_player_season_data.csv")

nba_final.head(50)
```
