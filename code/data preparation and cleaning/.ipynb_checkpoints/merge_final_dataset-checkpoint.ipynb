```python
# Loading cleaned datasets
from pathlib import Path
import pandas as pd

# Defining processed data directory
PROCESSED_DATA_DIR = Path.home() / "Desktop" / "NBA_Project" / "data" / "processed"

# Loading processed datasets
bio_info = pd.read_csv(PROCESSED_DATA_DIR / "bio_info_clean.csv")
player_career_info = pd.read_csv(PROCESSED_DATA_DIR / "player_career_info_clean.csv")
nba_performance = pd.read_csv(PROCESSED_DATA_DIR / "nba_performance_clean.csv")
team_standings = pd.read_csv(PROCESSED_DATA_DIR / "team_standings_clean.csv")
playoff_team_results = pd.read_csv(PROCESSED_DATA_DIR / "playoff_team_results_clean.csv")

# Checking dataset shapes
print("bio_info shape:", bio_info.shape)
print("player_career_info shape:", player_career_info.shape)
print("nba_performance shape:", nba_performance.shape)
print("team_standings shape:", team_standings.shape)
print("playoff_team_results shape:", playoff_team_results.shape)

# Displaying dataset columns
datasets = {
    "bio_info": bio_info,
    "player_career_info": player_career_info,
    "nba_performance": nba_performance,
    "team_standings": team_standings,
    "playoff_team_results": playoff_team_results
}

for name, df in datasets.items():
    print(f"\n{name} columns:")
    print(df.columns.tolist())
```

```python
############################################# merge bio info with career info ####################################
```

```python
# Merging player bio and career information

# Checking data quality before merge
print("Duplicates in player_career_info:", player_career_info["player_id"].duplicated().sum())
print("Missing player_id in bio_info:", bio_info["player_id"].isna().sum())
print("Missing player_name in bio_info:", bio_info["player_name"].isna().sum())

# Merging career information with player-season bio data
bio_career_info = bio_info.merge(
    player_career_info[
        [
            "player_id",
            "career_start_year",
            "career_end_year",
            "roster_status"
        ]
    ],
    on="player_id",
    how="left",
    validate="many_to_one"
)

# Checking merge results
print("bio_info shape:", bio_info.shape)
print("player_career_info shape:", player_career_info.shape)
print("bio_career_info shape:", bio_career_info.shape)

print("Missing career_start_year:", bio_career_info["career_start_year"].isna().sum())
print("Missing career_end_year:", bio_career_info["career_end_year"].isna().sum())
print("Missing roster_status:", bio_career_info["roster_status"].isna().sum())

# Reordering columns
bio_career_info = bio_career_info[[
    "year",
    "season",
    "player_id",
    "player_name",
    "team_id",
    "team",
    "age",
    "career_start_year",
    "career_end_year",
    "roster_status",
    "height_cm",
    "weight_kg",
    "college",
    "country",
    "draft_year",
    "draft_round",
    "draft_number",
    "games_played",
    "points",
    "rebounds",
    "assists",
    "net_rating",
    "usage_pct",
    "true_shooting_pct",
    "assist_pct",
    "offensive_rebound_pct",
    "defensive_rebound_pct"
]]

bio_career_info.head(50)
```

```python
############################################# merge bio/career data with performance data #############################################
```

```python
# Adding extra bio and career information to performance data

# Selecting additional variables from bio_career_info
bio_career_extra = bio_career_info[[
    "player_id",
    "year",
    "career_start_year",
    "career_end_year",
    "roster_status",
    "height_cm",
    "weight_kg",
    "college",
    "country",
    "draft_year",
    "draft_round",
    "draft_number",
    "net_rating",
    "usage_pct",
    "true_shooting_pct",
    "assist_pct",
    "offensive_rebound_pct",
    "defensive_rebound_pct"
]].copy()

# Keeping one row per player-season
bio_career_extra = bio_career_extra.drop_duplicates(
    subset=["player_id","year"]
)

# Merging performance data with bio and career information
nba_player_data = nba_performance.merge(
    bio_career_extra,
    on=["player_id","year"],
    how="left",
    validate="many_to_one"
)

# Checking merge result
print("nba_performance shape:", nba_performance.shape)
print("bio_career_extra shape:", bio_career_extra.shape)
print("nba_player_data shape:", nba_player_data.shape)

nba_player_data.head(50)
```

```python

############################################# show dataset columns #############################################
print("\nnba_player_data.columns:")
print(nba_player_data.columns.tolist())

print("\nteam_standings columns:")
print(team_standings.columns.tolist())

print("\nplayoff_team_results columns:")
print(playoff_team_results.columns.tolist())
```

```python
# Merging player data with team standings data

# Selecting additional team variables
team_standings_extra = team_standings[[
    "year",
    "team_id",
    "team_full_name",
    "conference",
    "division",
    "team_wins",
    "team_losses",
    "team_win_pct",
    "team_points_per_game",
    "team_opp_points_per_game",
    "team_point_diff_per_game"
]].copy()

# Keeping one row per team-season
team_standings_extra = team_standings_extra.drop_duplicates(
    subset=["year","team_id"]
)

# Merging player data with team information
nba_player_team_data = nba_player_data.merge(
    team_standings_extra,
    on=["year","team_id"],
    how="left",
    validate="many_to_one"
)

# Checking merge result
print("nba_player_data shape:", nba_player_data.shape)
print("team_standings_extra shape:", team_standings_extra.shape)
print("nba_player_team_data shape:", nba_player_team_data.shape)

print("Missing team_full_name:", nba_player_team_data["team_full_name"].isna().sum())
print("Missing conference:", nba_player_team_data["conference"].isna().sum())
print("Missing team_wins:", nba_player_team_data["team_wins"].isna().sum())

nba_player_team_data.head(50)
```

```python

############################################# show dataset columns #############################################

print("\nnba_player_team_data columns:")
print(nba_player_team_data.columns.tolist())

print("\nplayoff_team_results columns:")
print(playoff_team_results.columns.tolist())
```

```python
# Preparing playoff data and merging with player-team data
from difflib import get_close_matches

# Selecting useful playoff variables
playoff_team_extra = playoff_team_results[[
    "year",
    "season",
    "team_full_name",
    "playoff_seed",
    "actual_playoff_team",
    "top_8_conference_team",
    "playoff_stage",
    "playoff_finish_rank",
    "reached_conference_semifinals",
    "reached_conference_finals",
    "reached_finals",
    "won_nba_title"
]].copy()

# Cleaning merge keys
nba_player_team_data["year"] = pd.to_numeric(
    nba_player_team_data["year"],
    errors="coerce"
).astype("Int64")

playoff_team_extra["year"] = pd.to_numeric(
    playoff_team_extra["year"],
    errors="coerce"
).astype("Int64")

for df in [nba_player_team_data, playoff_team_extra]:
    df["season"] = df["season"].astype(str).str.strip()
    df["team_full_name"] = (
        df["team_full_name"]
        .astype(str)
        .str.replace("\xa0"," ",regex=False)
        .str.strip()
    )

# Correcting Clippers naming difference
team_names_available = nba_player_team_data[
    ["year","season","team_full_name"]
].drop_duplicates()

def fix_clippers_name(row):
    same_season_names = team_names_available[
        (team_names_available["year"] == row["year"]) &
        (team_names_available["season"] == row["season"])
    ]["team_full_name"].tolist()

    if row["team_full_name"] == "Los Angeles Clippers" and "LA Clippers" in same_season_names:
        return "LA Clippers"

    return row["team_full_name"]

playoff_team_extra["team_full_name_corrected"] = playoff_team_extra.apply(
    fix_clippers_name,
    axis=1
)

name_correction_check = playoff_team_extra[
    playoff_team_extra["team_full_name"] != playoff_team_extra["team_full_name_corrected"]
][[
    "year",
    "season",
    "team_full_name",
    "team_full_name_corrected"
]].sort_values(["year","team_full_name"])

print("Team name corrections:")
display(name_correction_check)

playoff_team_extra["team_full_name"] = playoff_team_extra["team_full_name_corrected"]
playoff_team_extra = playoff_team_extra.drop(columns="team_full_name_corrected")

# Checking duplicate playoff teams
duplicate_playoff_teams = playoff_team_extra[
    playoff_team_extra.duplicated(
        subset=["year","season","team_full_name"],
        keep=False
    )
].sort_values(["year","season","team_full_name"])

print("Duplicate playoff teams after correction:")
display(duplicate_playoff_teams)

playoff_team_extra = playoff_team_extra.drop_duplicates(
    subset=["year","season","team_full_name"]
)

# Checking playoff team count before merge
playoff_team_extra_count = (
    playoff_team_extra
    .groupby(["year","season"])["team_full_name"]
    .nunique()
    .reset_index()
    .rename(columns={
        "team_full_name":"number_of_playoff_teams_in_playoff_table"
    })
)

print("Number of playoff teams before merge:")
display(playoff_team_extra_count)

print("Seasons with not exactly 16 playoff teams:")
display(
    playoff_team_extra_count[
        playoff_team_extra_count["number_of_playoff_teams_in_playoff_table"] != 16
    ]
)

# Checking unmatched playoff team names
missing_playoff_team_matches = playoff_team_extra.merge(
    team_names_available,
    on=["year","season","team_full_name"],
    how="left",
    indicator=True
)

missing_playoff_team_matches = missing_playoff_team_matches[
    missing_playoff_team_matches["_merge"] == "left_only"
].drop(columns="_merge")

def find_possible_match(row):
    same_season_names = team_names_available[
        (team_names_available["year"] == row["year"]) &
        (team_names_available["season"] == row["season"])
    ]["team_full_name"].tolist()

    return get_close_matches(
        row["team_full_name"],
        same_season_names,
        n=3,
        cutoff=0.5
    )

if len(missing_playoff_team_matches) > 0:
    missing_playoff_team_matches["possible_matches"] = missing_playoff_team_matches.apply(
        find_possible_match,
        axis=1
    )

print("Playoff team names missing before merge:")
display(
    missing_playoff_team_matches.sort_values(
        ["year","team_full_name"]
    )
)

# Merging playoff results
nba_final = nba_player_team_data.merge(
    playoff_team_extra,
    on=["year","season","team_full_name"],
    how="left",
    validate="many_to_one"
)

# Filling missing playoff information
nba_final["actual_playoff_team"] = nba_final["actual_playoff_team"].fillna(0).astype(int)
nba_final["top_8_conference_team"] = nba_final["top_8_conference_team"].fillna(0).astype(int)

playoff_flag_columns = [
    "reached_conference_semifinals",
    "reached_conference_finals",
    "reached_finals",
    "won_nba_title"
]

for col in playoff_flag_columns:
    nba_final[col] = nba_final[col].fillna(0).astype(int)

# Checking playoff team count
playoff_team_count_check = (
    nba_final[nba_final["actual_playoff_team"] == 1]
    .groupby(["year","season"])["team_full_name"]
    .nunique()
    .reset_index()
    .rename(columns={
        "team_full_name":"number_of_playoff_teams"
    })
)

print("Number of playoff teams per season:")
display(playoff_team_count_check)

print("Seasons with not exactly 16 playoff teams:")
display(
    playoff_team_count_check[
        playoff_team_count_check["number_of_playoff_teams"] != 16
    ]
)
```

```python
# Cleaning playoff variables for player-team context

# Filling missing playoff information for non-playoff teams
nba_final["top_8_conference_team"] = nba_final["top_8_conference_team"].fillna(0).astype(int)

nba_final["playoff_seed"] = nba_final["playoff_seed"].fillna(0).astype(int)

nba_final["playoff_stage"] = nba_final["playoff_stage"].fillna("No Playoffs")

# Adding playoff finish rank
nba_final["playoff_finish_rank"] = nba_final["playoff_finish_rank"].fillna(6).astype(int)

# Filling playoff indicators
playoff_flag_columns = [
    "reached_conference_semifinals",
    "reached_conference_finals",
    "reached_finals",
    "won_nba_title"
]

for col in playoff_flag_columns:
    nba_final[col] = nba_final[col].fillna(0).astype(int)
```

```python
############################################# reorder and save final dataset #####################################
```

```python
# Creating player experience variable
nba_final["player_experience_years"] = (
    nba_final["year"] - nba_final["career_start_year"]
)

# Filling non-playoff teams indicator
nba_final["actual_playoff_team"] = nba_final["actual_playoff_team"].fillna(0).astype(int)

# Removing unused playoff outcome variables
playoff_columns_to_remove = [
    "top_8_conference_team",
    "playoff_seed",
    "playoff_stage",
    "playoff_finish_rank",
    "reached_conference_semifinals",
    "reached_conference_finals",
    "reached_finals",
    "won_nba_title"
]

nba_final = nba_final.drop(
    columns=playoff_columns_to_remove,
    errors="ignore"
)

# Reordering columns
preferred_order = [
    "year",
    "season",

    "player_id",
    "player_name",

    "age",
    "career_start_year",
    "player_experience_years",
    "career_end_year",
    "roster_status",

    "height_cm",
    "weight_kg",
    "college",
    "country",

    "draft_year",
    "draft_round",
    "draft_number",

    "games_played",
    "minutes",
    "wins",
    "losses",
    "win_pct",

    "points",
    "field_goals_made",
    "field_goal_attempts",
    "field_goal_pct",
    "three_points_made",
    "three_point_attempts",
    "three_point_pct",
    "free_throws_made",
    "free_throw_attempts",
    "free_throw_pct",

    "offensive_rebounds",
    "defensive_rebounds",
    "rebounds",

    "assists",
    "turnovers",

    "steals",
    "blocks",
    "blocks_against",

    "personal_fouls",
    "personal_fouls_drawn",

    "plus_minus",
    "nba_fantasy_points",
    "double_doubles",
    "triple_doubles",

    "net_rating",
    "usage_pct",
    "true_shooting_pct",
    "assist_pct",
    "offensive_rebound_pct",
    "defensive_rebound_pct",

    "team_id",
    "team",
    "team_full_name",
    "conference",
    "division",
    "team_count",

    "team_wins",
    "team_losses",
    "team_win_pct",
    "team_points_per_game",
    "team_opp_points_per_game",
    "team_point_diff_per_game",

    "actual_playoff_team"
]

preferred_order = [
    col for col in preferred_order
    if col in nba_final.columns
]

remaining_columns = [
    col for col in nba_final.columns
    if col not in preferred_order
]

nba_final = nba_final[
    preferred_order + remaining_columns
]

print("Final dataset shape:", nba_final.shape)

display(nba_final.head(50))
```
