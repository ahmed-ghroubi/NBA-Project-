```python
############################################# load raw player info #################

import pandas as pd
from nba_api.stats.endpoints import CommonAllPlayers

# Load all historical NBA players from nba_api
players_info = CommonAllPlayers(
    is_only_current_season=0,
    season="2025-26",
    league_id="00"
).get_data_frames()[0]

# Save raw player info data
players_info.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\raw\players_info_raw.csv",
    index=False
)

# Check raw data
print("Original players_info shape:", players_info.shape)

players_info.head(50)
```

```python
############################################# clean player career info #################

# Work on a copy of the raw player info data
players_info_clean = players_info.copy()

# Convert FROM_YEAR and TO_YEAR to numeric values
players_info_clean["FROM_YEAR"] = pd.to_numeric(
    players_info_clean["FROM_YEAR"],
    errors="coerce"
)

players_info_clean["TO_YEAR"] = pd.to_numeric(
    players_info_clean["TO_YEAR"],
    errors="coerce"
)

# Keep only players who were active from the 1997-98 season onward
# A player can start before 1997-98, but still be relevant if TO_YEAR >= 1997
# TO_YEAR is the starting year of the last NBA season, e.g. 1997 means 1997-98
players_info_filtered = players_info_clean[
    players_info_clean["TO_YEAR"] >= 1997
].copy()

# Keep only the important columns for the project
player_career_info = players_info_filtered[
    [
        "PERSON_ID",
        "DISPLAY_FIRST_LAST",
        "FROM_YEAR",
        "TO_YEAR",
        "ROSTERSTATUS",
        
    ]
].copy()

# Rename columns to cleaner names
player_career_info = player_career_info.rename(columns={
    "PERSON_ID": "player_id",
    "DISPLAY_FIRST_LAST": "player_name",
    "FROM_YEAR": "career_start_year",
    "TO_YEAR": "career_end_year",
    "ROSTERSTATUS": "roster_status",
  
})

# Convert career years to integer format
player_career_info["career_start_year"] = player_career_info["career_start_year"].astype("Int64")

# Add +1 to career_end_year because TO_YEAR describes the starting year of the last NBA season
# Example: 2000 means season 2000-01, so the ending year is 2001
player_career_info["career_end_year"] = (
    player_career_info["career_end_year"].astype("Int64") + 1
)

# Remove .0 from roster_status
player_career_info["roster_status"] = player_career_info["roster_status"].astype("Int64")

# Correct known same-name issue: Mark Jones, player_id 2891
# The correct NBA career start year is 2004
player_career_info.loc[
    player_career_info["player_id"] == 2891,
    "career_start_year"
] = 2004

# Reorder columns
player_career_info = player_career_info[
    [
        "player_id",
        "player_name",
        "career_start_year",
        "career_end_year",
        "roster_status",
       
    ]
]

# Save cleaned player career info data
player_career_info.to_csv(
    r"C:\Users\ahmed\Desktop\NBA_Project\data\processed\player_career_info_clean.csv",
    index=False
)

# Check result
print("Original players_info rows:", players_info.shape[0])
print("Filtered players rows:", players_info_filtered.shape[0])
print("Final player_career_info shape:", player_career_info.shape)

player_career_info.head()
```
