```python
# Loading raw player performance data
import pandas as pd
import time
from pathlib import Path
from nba_api.stats.endpoints import LeagueDashPlayerStats

all_seasons = []

def make_season_string(end_year):
    return f"{end_year - 1}-{str(end_year)[-2:]}"

# Downloading all seasons
for end_year in range(1997, 2027):
    season = make_season_string(end_year)
    print("Loading performance stats:", season)
    try:
        stats = LeagueDashPlayerStats(
            season=season,
            season_type_all_star="Regular Season",
            per_mode_detailed="Totals",
            timeout=60
        )
        season_df = stats.get_data_frames()[0]
        season_df["Season"] = season
        season_df["Year"] = end_year
        all_seasons.append(season_df)
        time.sleep(2)
    except Exception as e:
        print("Could not load:", season, e)

# Combining all seasons
if all_seasons:
    nba_stats_official = pd.concat(all_seasons, ignore_index=True)

# Saving raw data
raw_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)

output_file = raw_dir / "nba_player_performance_raw.csv"
nba_stats_official.to_csv(output_file, index=False)

print("Saved raw player performance data to:", output_file)
print("Raw performance data shape:", nba_stats_official.shape)

nba_stats_official.head()
```

```python
# Cleaning player performance data

rank_columns = [col for col in nba_stats_official.columns if col.endswith("_RANK")]
nba_performance = nba_stats_official.drop(columns=rank_columns).copy()

# Selecting useful variables
nba_performance = nba_performance[[
    "Year","Season",
    "PLAYER_ID","PLAYER_NAME",
    "TEAM_ID","TEAM_ABBREVIATION","TEAM_COUNT",
    "AGE","GP","MIN",
    "W","L","W_PCT",
    "PTS","FGM","FGA","FG_PCT",
    "FG3M","FG3A","FG3_PCT",
    "FTM","FTA","FT_PCT",
    "OREB","DREB","REB",
    "AST","TOV",
    "STL","BLK","BLKA",
    "PF","PFD",
    "PLUS_MINUS","NBA_FANTASY_PTS",
    "DD2","TD3"
]].copy()

# Renaming columns
nba_performance = nba_performance.rename(columns={
    "Year":"year",
    "Season":"season",
    "PLAYER_ID":"player_id",
    "PLAYER_NAME":"player_name",
    "TEAM_ID":"team_id",
    "TEAM_ABBREVIATION":"team",
    "TEAM_COUNT":"team_count",
    "AGE":"age",
    "GP":"games_played",
    "MIN":"minutes",
    "W":"wins",
    "L":"losses",
    "W_PCT":"win_pct",
    "PTS":"points",
    "FGM":"field_goals_made",
    "FGA":"field_goal_attempts",
    "FG_PCT":"field_goal_pct",
    "FG3M":"three_points_made",
    "FG3A":"three_point_attempts",
    "FG3_PCT":"three_point_pct",
    "FTM":"free_throws_made",
    "FTA":"free_throw_attempts",
    "FT_PCT":"free_throw_pct",
    "OREB":"offensive_rebounds",
    "DREB":"defensive_rebounds",
    "REB":"rebounds",
    "AST":"assists",
    "TOV":"turnovers",
    "STL":"steals",
    "BLK":"blocks",
    "BLKA":"blocks_against",
    "PF":"personal_fouls",
    "PFD":"personal_fouls_drawn",
    "PLUS_MINUS":"plus_minus",
    "NBA_FANTASY_PTS":"nba_fantasy_points",
    "DD2":"double_doubles",
    "TD3":"triple_doubles"
})

# Creating additional shooting variables
nba_performance["two_points_made"] = nba_performance["field_goals_made"] - nba_performance["three_points_made"]
nba_performance["two_point_attempts"] = nba_performance["field_goal_attempts"] - nba_performance["three_point_attempts"]
nba_performance["two_point_pct"] = (nba_performance["two_points_made"] / nba_performance["two_point_attempts"]).fillna(0)

# Reordering columns
nba_performance = nba_performance[[
    "year",
    "season",
    "player_id",
    "player_name",
    "team_id",
    "team",
    "team_count",
    "age",
    "games_played",
    "minutes",
    "wins",
    "losses",
    "win_pct",
    "points",
    "field_goals_made",
    "field_goal_attempts",
    "field_goal_pct",
    "two_points_made",
    "two_point_attempts",
    "two_point_pct",
    "three_points_made",
    "three_point_attempts",
    "three_point_pct",
    "free_throws_made",
    "free_throw_attempts",
    "free_throw_pct",
    "offensive_rebounds",
    "defensive_rebounds",
    "rebounds",
    "assists",
    "turnovers",
    "steals",
    "blocks",
    "blocks_against",
    "personal_fouls",
    "personal_fouls_drawn",
    "plus_minus",
    "nba_fantasy_points",
    "double_doubles",
    "triple_doubles"
]].copy()

# Cleaning age format
nba_performance["age"] = pd.to_numeric(
    nba_performance["age"],
    errors="coerce"
).astype("Int64")

# Saving cleaned data
processed_dir = Path.home() / "Desktop" / "NBA_Project" / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

output_file = processed_dir / "nba_performance_clean.csv"
nba_performance.to_csv(output_file, index=False)

print("Saved cleaned performance data to:", output_file)
print("Original nba_stats_official shape:", nba_stats_official.shape)
print("Removed rank columns:", len(rank_columns))
print("Clean nba_performance shape:", nba_performance.shape)

nba_performance.head(50)
```
